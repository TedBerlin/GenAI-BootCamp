**Exercise 1: Bar Graph - Retail Sales**

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns # Used for setting a nice plot style

# Set a clean plot style
sns.set_style("whitegrid")

print("--- Exercise 1: Bar Graph ---")
```

```python
# 1. Create a fictional dataset
print("\n1. Creating fictional sales data...")
product_categories = ['Electronics', 'Clothing', 'Home Goods', 'Books', 'Groceries', 'Toys']
total_sales = [150000, 120000, 90000, 60000, 200000, 75000] # Fictional sales amounts

# Create a Pandas DataFrame for better organization (optional but good practice)
sales_data = pd.DataFrame({
    'Product Category': product_categories,
    'Total Sales': total_sales
})

print("Fictional sales data created:")
print(sales_data)
```

```python
# 2. Customize and Display the bar graph using Matplotlib
print("\n2. Generating and displaying the bar graph...")

plt.figure(figsize=(10, 6)) # Set the figure size for better readability

# Create the bar graph
# x-axis: product categories, y-axis: total sales amounts
# 'color' parameter can be a single color or a list of colors
# Using a single color for simplicity, but a palette could be used with seaborn
plt.bar(sales_data['Product Category'], sales_data['Total Sales'], color='skyblue')

# Customize the graph
plt.xlabel('Product Category', fontsize=12) # Label for the x-axis
plt.ylabel('Total Sales ($)', fontsize=12) # Label for the y-axis
plt.title('Total Sales by Product Category', fontsize=14) # Title of the graph

# Rotate x-axis labels if they overlap
plt.xticks(rotation=45, ha='right') # 'ha' stands for horizontal alignment

# Add grid for better readability (optional, but good for bar graphs)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Ensure layout is tight to prevent labels from being cut off
plt.tight_layout()

# Display the graph
plt.show()

print("\nBar graph displayed successfully.")
```

**Exercise 2: Hierarchical Indexing**


```python
import pandas as pd
import numpy as np

print("--- Exercice 2: Indexation Hiérarchique ---")

# 1. Créer un DataFrame Pandas avec indexation hiérarchique
print("\n1. Création d'un DataFrame avec indexation hiérarchique...")

# Données fictives pour simuler des températures dans différentes villes et pays
data = {
    'Country': ['Canada', 'Canada', 'Canada', 'Canada', 'Canada', 'Canada', 'Canada', 'Canada',
                'USA', 'USA', 'USA', 'USA', 'France', 'France', 'France', 'France'],
    'City': ['Toronto', 'Toronto', 'Montreal', 'Montreal', 'Vancouver', 'Vancouver', 'Toronto', 'Montreal',
             'New York', 'New York', 'Los Angeles', 'Los Angeles', 'Paris', 'Paris', 'Lyon', 'Lyon'],
    'Date': pd.to_datetime([
        '2023-07-10', '2023-07-11', '2023-07-10', '2023-07-11', '2023-07-10', '2023-07-11', '2023-08-01', '2023-08-02',
        '2023-07-10', '2023-07-11', '2023-07-10', '2023-07-11', '2023-07-10', '2023-07-11', '2023-07-10', '2023-07-11'
    ]),
    'Temperature': [
        28, 31, 29, 30, 25, 27, 32, 30, # Canada
        35, 36, 28, 30, 22, 24, 26, 28  # USA, France
    ]
}

df = pd.DataFrame(data)
```

```python
# Définir l'index hiérarchique
df_indexed = df.set_index(['Country', 'City', 'Date'])

print("\nDataFrame original (non indexé):")
print(df)
print("\nDataFrame avec index hiérarchique:")
print(df_indexed)
print(f"\nType d'index: {type(df_indexed.index)}")
```

```python

# 2. Écrire du code Python pour filtrer le dataset
print("\n2. Filtrage du dataset: Température > 30°C pour le Canada...")

# Filtrer directement sur l'index hiérarchique et la colonne 'Temperature'
# Utilisation de .loc pour un filtrage efficace sur l'index et les colonnes
filtered_data = df_indexed.loc[(('Canada'), slice(None), slice(None))] # Sélectionne toutes les villes et dates au Canada
filtered_data = filtered_data[filtered_data['Temperature'] > 30]

print("\nJours où la température a dépassé 30°C au Canada:")
print(filtered_data)

```

**Indexation Hiérarchique**


L'indexation hiérarchique (également appelée MultiIndex dans Pandas) est une fonctionnalité puissante qui permet de gérer des données avec plusieurs niveaux d'indexation. Au lieu d'avoir un seul index simple (comme des numéros de ligne), vous pouvez avoir deux, trois, ou même plus de niveaux d'index, créant une structure de données plus riche et plus organisée.

**Comment l'indexation hiérarchique simplifie l'analyse de données à plusieurs niveaux :**

1. Organisation et Lisibilité :

* Elle permet d'organiser des données complexes de manière intuitive, comme des données de séries temporelles pour plusieurs entités (par exemple, Pays > Ville > Date > Température). Cela rend le DataFrame plus facile à lire et à comprendre, car les relations hiérarchiques sont visuellement claires.

2. Sélection et Filtrage Simplifiés :

* L'accès aux données devient beaucoup plus simple et plus puissant. Vous pouvez sélectionner des données à n'importe quel niveau de la hiérarchie en utilisant la méthode .loc[] de Pandas.

* Exemple : Au lieu de filtrer d'abord par df[df['Country'] == 'Canada'] puis df[df['City'] == 'Toronto'], vous pouvez directement utiliser df_indexed.loc['Canada', 'Toronto'] pour accéder aux données de Toronto au Canada.

* Vous pouvez également utiliser des slice(None) pour sélectionner tous les éléments à un certain niveau, comme df_indexed.loc[('Canada', slice(None), '2023-07-11')] pour toutes les villes du Canada à une date spécifique.


3. Opérations de Groupe Efficaces :

* Les opérations de regroupement (groupby()) deviennent plus naturelles et performantes. Vous pouvez facilement effectuer des agrégations à n'importe quel niveau de la hiérarchie sans avoir à spécifier explicitement les colonnes de regroupement.

* Exemple : Calculer la température moyenne par pays, puis par ville, est plus direct avec un MultiIndex.


4. Réduction de la Répétition des Données :

* Dans un DataFrame sans index hiérarchique, les valeurs des colonnes "Country" et "City" seraient répétées pour chaque ligne. Avec un MultiIndex, ces valeurs sont stockées de manière plus efficace dans l'index, ce qui peut réduire la redondance visuelle et parfois la consommation de mémoire pour de très grands datasets.


5. Préparation pour l'Analyse Multidimensionnelle :

* Elle jette les bases d'analyses plus complexes, comme les tableaux croisés dynamiques (pivot_table) ou les modèles statistiques qui bénéficient d'une structure de données bien définie.

En résumé, l'indexation hiérarchique transforme un tableau plat en une structure de données plus riche et plus intuitive, ce qui simplifie considérablement la manipulation, le filtrage et l'analyse des données qui ont une organisation naturelle à plusieurs niveaux.

**Exercise 3: Advanced Filtering with Hierarchical Indices**


```python
import pandas as pd
import numpy as np

print("--- Exercice 3: Filtrage Avancé avec Indices Hiérarchiques ---")

# 1. Créer un DataFrame avec indexation hiérarchique
# Nous allons simuler un DataFrame qui pourrait provenir d'un "Exercice 4 des exercices XP",
# avec des départements, des employés et leurs salaires.
print("\n1. Création d'un DataFrame avec indexation hiérarchique (Département, Employé)...")

data = {
    'Department': ['HR', 'HR', 'HR', 'IT', 'IT', 'IT', 'IT', 'Sales', 'Sales', 'Sales'],
    'EmployeeID': ['E001', 'E002', 'E003', 'E004', 'E005', 'E006', 'E007', 'E008', 'E009', 'E010'],
    'Salary': [45000, 60000, 52000, 70000, 48000, 95000, 55000, 62000, 49000, 80000],
    'YearsExperience': [2, 5, 3, 8, 2, 10, 4, 6, 1, 7]
}

df = pd.DataFrame(data)

# Définir l'index hiérarchique sur 'Department' et 'EmployeeID'
df_indexed = df.set_index(['Department', 'EmployeeID'])

print("\nDataFrame original avec index hiérarchique:")
print(df_indexed)
```

```python
# 2. Filtrer et afficher les lignes où 'Salary' est supérieur à 50,000 dans chaque 'Department'
print("\n2. Filtrage du dataset: Salaires > 50,000 par Département...")

# Pour filtrer les lignes où 'Salary' est > 50,000,
# on peut directement appliquer la condition sur la colonne 'Salary' du DataFrame indexé.
# Pandas gérera cela efficacement pour chaque niveau de l'index.
filtered_salaries = df_indexed[df_indexed['Salary'] > 50000]

print("\nEmployés avec un salaire supérieur à 50,000 par département:")
print(filtered_salaries)

```

**Filtrage Avancé avec Indices Hiérarchiques**


L'indexation hiérarchique, ou MultiIndex dans Pandas, est particulièrement utile pour le filtrage avancé de données qui ont une structure imbriquée ou des relations de groupe naturelles.

**Comment l'indexation hiérarchique facilite un filtrage efficace :**

Dans l'exemple de l'exercice, nous avons un DataFrame indexé par Department et EmployeeID. Nous voulons filtrer les employés dont le Salary est supérieur à 50 000, et ce, "dans chaque département".

1. Application Directe de Conditions sur les Colonnes :

* Lorsque vous avez un DataFrame avec un MultiIndex, vous pouvez appliquer des conditions directement sur les colonnes non-indexées (comme Salary dans notre cas) de la même manière que pour un DataFrame standard.

* df_indexed[df_indexed['Salary'] > 50000] : Pandas applique cette condition à chaque ligne du DataFrame, quelle que soit la hiérarchie de l'index. Le résultat est un nouveau DataFrame qui conserve la structure hiérarchique de l'index, mais ne contient que les lignes qui satisfont la condition.

* Cela simplifie l'opération car vous n'avez pas besoin de boucler manuellement sur chaque département pour appliquer le filtre. Pandas gère l'efficacité sous le capot.


2. Clarté et Intuitivité :

* Le résultat du filtrage (filtered_salaries) est un DataFrame qui conserve les niveaux d'index Department et EmployeeID. Cela rend le résultat immédiatement compréhensible : vous voyez clairement à quel département et à quel employé chaque enregistrement de salaire filtré appartient.

* Si nous n'avions pas d'index hiérarchique, nous aurions un DataFrame plat et il faudrait peut-être regrouper par Department après le filtrage pour obtenir la même clarté, ou utiliser des requêtes plus complexes.

3. Filtrage Combiné (Index et Colonnes) :

* L'indexation hiérarchique devient encore plus puissante lorsque vous combinez le filtrage sur les niveaux de l'index avec le filtrage sur les colonnes.

* Exemple : si nous voulions les salaires > 50 000 uniquement pour le département 'IT', nous pourrions faire :

df_indexed.loc['IT'][df_indexed.loc['IT']['Salary'] > 50000]

Ou plus élégamment :

df_indexed.loc[('IT', slice(None)), :][df_indexed.loc[('IT', slice(None)), :]['Salary'] > 50000]

Cela montre comment .loc permet une sélection précise sur les niveaux de l'index avant d'appliquer d'autres conditions.

En résumé, l'indexation hiérarchique ne se contente pas d'organiser les données ; elle fournit un cadre puissant et intuitif pour effectuer des opérations de sélection et de filtrage complexes à travers plusieurs niveaux de catégories, rendant le code plus propre et l'analyse plus efficace.
