**Prompting for Control & Output Quality**


**Step 1: Craft a Prompt That Controls Output Tone, Format, and Length**

Prompt:
"Rewrite the following text using a friendly and clear tone, using bullet points to organize the information, paraphrasing rather than quoting directly, and staying under 75 words total:

'Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications.’"


**Step 2: Evaluate the Output**

Generated Output:

* Always use the approved VPN for remote access.
* Avoid unsecured connections.
* Do not use personal devices without protection.
* Protect sensitive data and communications.


**Evaluation:**

* **Relevance:** Yes, the output stays true to the message of the source content.
* **Clarity**: Yes, the message is easier to understand for non-technical staff.
* **Structure**: Yes, the information is formatted into clear bullet points.
* **Tone**: Yes, the language is appropriate for internal employee communication.
* **Length**: Yes, the output stays under 75 words.
* **Factual Accuracy:** Yes, no critical security details were invented or removed.


**Step 3: Detect and Mitigate Hallucinations**

If the model adds extra content, I would rewrite the prompt as follows:

"Only paraphrase the content in the input text. Do not add new recommendations or technologies. »


**Step 4: Paraphrasing Deep Dive**

Prompt:
"Rewrite the policy paragraph for a junior intern audience using plain language and short phrases. Use no more than 4 bullet points. Avoid any corporate or legal jargon. Keep the tone supportive and informative. »

Generated Output:
* Use the approved VPN for remote access.
* Avoid unsecured connections.
* Don't use personal devices without protection.
* Protect sensitive data.


**Step 5: Quote Extraction Variant**

Prompt:
"Extract one direct quote from the original text that best captures the core security policy."

**Generated Output:**
"Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications. »

**Responses to the questions:**

**When is quoting more appropriate than paraphrasing?** : Quoting is more appropriate in formal internal communications, such as official policy updates or legal communications, where the precision of the original language is crucial.

**When might quoting pose a risk?** : Quoting can pose a risk if it is taken out of context or if it is misunderstood by the target audience, especially if the language is too technical or legalistic for a non-specialized audience.
