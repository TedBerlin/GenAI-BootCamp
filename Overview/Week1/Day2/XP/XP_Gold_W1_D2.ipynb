```python
# Exercise 1: Birthday Look-up

def birthday_lookup():
    """
    Manages a dictionary of birthdays and allows the user to look up birthdays by name.
    """
    # Create a dictionary to store birthdays
    # Key: Person's Name (str), Value: Birthday (str in "YYYY/MM/DD" format)
    birthdays = {
        "Alice": "1990/05/15",
        "Bob": "1988/11/22",
        "Charlie": "1992/03/01",
        "Diana": "1985/07/30",
        "Eve": "1995/09/10"
    }

    # Print a welcome message
    print("--- Exercise 1: Birthday Look-up ---")
    print("Welcome to the Birthday Look-up tool!")
    print("You can look up the birthdays of the following people:")

    # Display the names available for lookup
    for name in birthdays.keys():
        print(f"- {name}")

    # Ask the user for a person's name
    person_name = input("\nPlease enter the name of the person whose birthday you want to look up: ").strip()

    # Get the birthday of the name provided by the user
    # Use .get() method to avoid KeyError if the name is not found, returning None by default
    birthday = birthdays.get(person_name)

    # Print out the birthday with a nicely-formatted message
    if birthday:
        print(f"\n{person_name}'s birthday is on: {birthday}")
    else:
        print(f"\nSorry, I don't have the birthday for '{person_name}' in my list.")

# Call the function to run the program
if __name__ == "__main__":
    birthday_lookup()
```

```python
# Exercise 2: Birthdays Advanced

birthdays = {
    "Alice": "15/04/1990",
    "Bob": "23/07/1985",
    "Charlie": "10/12/1978",
    "Diana": "30/01/1995",
    "Eve": "05/11/1982"
}

def print_birthday_info(birthday_dict):
    # Print all names in the dictionary
    print("Here are all the names in our birthday records:")
    for name in birthday_dict:
        print(f"- {name}")

    # Ask user for a name
    user_input = input("\nEnter a person's name to get their birthday: ").strip()

    # Check if the name exists in the dictionary
    if user_input in birthday_dict:
        print(f"\n{user_input}'s birthday is {birthday_dict[user_input]}")
    else:
        print(f"\nSorry, we don't have the birthday information for {user_input}")

# Call the function with our birthday dictionary
print_birthday_info(birthdays)
```

```python
# Exercise 3: Check the Index

# Define the list of names as specified in the instructions
names = ['Samus', 'Cortana', 'V', 'Link', 'Mario', 'Cortana', 'Samus']

print("--- Exercise 3: Check the Index ---")
print(f"Here is our list of names: {names}")

# Ask the user for their name
user_name = input("Please enter a name to check its first occurrence in the list: ").strip()

# Check if the user's name is in the list
if user_name in names:
    # If the name is found, use the .index() method to get the index of its first occurrence
    first_occurrence_index = names.index(user_name)
    print(f"The name '{user_name}' was found at the first occurrence index: {first_occurrence_index}")
else:
    # If the name is not found, inform the user
    print(f"The name '{user_name}' was not found in the list.")
```

```python
# Exercise 4: Double Dice

import random

def throw_dice():
    """
    Simulates the rolling of a single dice.
    Returns an integer between 1 and 6 (inclusive).
    """
    return random.randint(1, 6)

def throw_until_doubles():
    """
    Keeps throwing two dice until they both land on the same number (doubles).
    Returns the number of times it threw the dice in total.
    """
    num_throws = 0
    while True:
        num_throws += 1
        die1 = throw_dice()
        die2 = throw_dice()
        # print(f"  Roll {num_throws}: ({die1}, {die2})") # Uncomment for detailed roll output
        if die1 == die2:
            # print(f"  Doubles reached in {num_throws} throws!") # Uncomment for detailed roll output
            return num_throws

def main():
    """
    Main function to simulate throwing doubles 100 times,
    store the results, and print statistics.
    """
    print("--- Exercise 4: Double Dice ---")
    print("Simulating dice throws to reach doubles 100 times...")

    results_of_throws = [] # Collection to store the number of throws for each doubles event
    num_simulations = 100

    for i in range(num_simulations):
        # Call throw_until_doubles and store the result
        throws_for_doubles = throw_until_doubles()
        results_of_throws.append(throws_for_doubles)
        # print(f"Simulation {i+1}: Doubles in {throws_for_doubles} throws.") # Uncomment to see each simulation result

    # Calculate total throws
    total_throws_all_simulations = sum(results_of_throws)

    # Calculate average throws to reach doubles
    average_throws = total_throws_all_simulations / num_simulations

    # Print the results
    print(f"\nTotal throws to reach {num_simulations} doubles: {total_throws_all_simulations}")
    print(f"Average throws to reach doubles: {average_throws:.2f}")

# Call the main function to run the simulation
if __name__ == "__main__":
    main()
```
