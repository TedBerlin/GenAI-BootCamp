```python
# Exercise 1: Geometry

import math

class Circle:
    """
    A class to represent a circle with a given radius.
    Provides methods to compute its perimeter and area, and print its definition.
    """
    def __init__(self, radius=1.0):
        """
        Initializes a Circle object with a given radius.

        Args:
            radius (float): The radius of the circle. Defaults to 1.0.
                            Must be a non-negative number.
        """
        if not isinstance(radius, (int, float)):
            raise TypeError("Radius must be a number.")
        if radius < 0:
            raise ValueError("Radius cannot be negative.")
        self.radius = float(radius)

    def perimeter(self):
        """
        Computes and returns the perimeter (circumference) of the circle.
        Formula: 2 * pi * radius
        """
        return 2 * math.pi * self.radius

    def area(self):
        """
        Computes and returns the area of the circle.
        Formula: pi * radius^2
        """
        return math.pi * (self.radius ** 2)

    def print_definition(self):
        """
        Prints the geometrical definition of a circle.
        """
        print("--- Geometrical Definition of a Circle ---")
        print("A circle is a round plane figure whose boundary (the circumference)")
        print("consists of points equidistant from a fixed central point.")
        print("------------------------------------------")

# --- Example Usage ---
if __name__ == "__main__":
    print("--- Exercise 1: Geometry - Circle Class ---")

    # Create a circle with default radius
    circle1 = Circle()
    print("\n--- Circle 1 (Default Radius) ---")
    print(f"Radius: {circle1.radius}")
    print(f"Perimeter: {circle1.perimeter():.2f}")
    print(f"Area: {circle1.area():.2f}")
    circle1.print_definition()

    # Create a circle with a specific radius
    circle2 = Circle(radius=5.0)
    print("\n--- Circle 2 (Radius = 5.0) ---")
    print(f"Radius: {circle2.radius}")
    print(f"Perimeter: {circle2.perimeter():.2f}")
    print(f"Area: {circle2.area():.2f}")

    # Create a circle with a different radius
    circle3 = Circle(radius=10)
    print("\n--- Circle 3 (Radius = 10) ---")
    print(f"Radius: {circle3.radius}")
    print(f"Perimeter: {circle3.perimeter():.2f}")
    print(f"Area: {circle3.area():.2f}")

    # Example with invalid input (will raise an error)
    # try:
    #     circle_invalid = Circle(radius=-2.0)
    # except ValueError as e:
    #     print(f"\nError creating circle: {e}")
    # try:
    #     circle_invalid_type = Circle(radius="abc")
    # except TypeError as e:
    #     print(f"\nError creating circle: {e}")
```

```python
# Exercise 2: Custom List Class

import random

class MyList:
    """
    A class to encapsulate list operations.
    It receives a list of letters and provides methods for reversing, sorting,
    and generating a new list with random numbers of the same length.
    """
    def __init__(self, letters_list):
        """
        Initializes a MyList object with a list of letters.

        Args:
            letters_list (list): A list of characters (letters).
        """
        if not isinstance(letters_list, list):
            raise TypeError("Input must be a list.")
        # Optionally, you could add a check to ensure all elements are single characters/letters
        self.original_list = list(letters_list) # Store a copy to preserve original

    def get_reversed_list(self):
        """
        Returns a new list with the elements of the original list in reversed order.
        """
        return self.original_list[::-1] # Slicing creates a reversed copy

    def get_sorted_list(self):
        """
        Returns a new list with the elements of the original list in sorted order.
        """
        return sorted(self.original_list) # sorted() returns a new sorted list

    def generate_random_numbers_list(self, min_val=0, max_val=100):
        """
        Bonus: Generates a second list with the same length as the original list,
        constructed with random numbers within a specified range.

        Args:
            min_val (int): The minimum value for the random numbers (inclusive).
            max_val (int): The maximum value for the random numbers (inclusive).

        Returns:
            list: A new list of random integers.
        """
        if not self.original_list:
            return [] # Return empty list if original is empty

        # Use list comprehension to generate random numbers
        random_numbers_list = [random.randint(min_val, max_val) for _ in range(len(self.original_list))]
        return random_numbers_list

# --- Example Usage ---
if __name__ == "__main__":
    print("--- Exercise 2: Custom List Class ---")

    # Define a list of letters
    my_letters = ['c', 'a', 'z', 'b', 'x', 'd']
    print(f"\nOriginal list: {my_letters}")

    # Create an instance of MyList
    my_list_instance = MyList(my_letters)

    # Test get_reversed_list method
    reversed_list = my_list_instance.get_reversed_list()
    print(f"Reversed list: {reversed_list}")

    # Test get_sorted_list method
    sorted_list = my_list_instance.get_sorted_list()
    print(f"Sorted list: {sorted_list}")

    # Test generate_random_numbers_list method (Bonus)
    random_list = my_list_instance.generate_random_numbers_list(min_val=1, max_val=50)
    print(f"Generated random numbers list (same length): {random_list}")

    # Test with an empty list
    empty_list_instance = MyList([])
    print(f"\nOriginal empty list: {empty_list_instance.original_list}")
    print(f"Reversed empty list: {empty_list_instance.get_reversed_list()}")
    print(f"Sorted empty list: {empty_list_instance.get_sorted_list()}")
    print(f"Generated random numbers for empty list: {empty_list_instance.generate_random_numbers_list()}")

    # Test with a different list
    another_list = ['apple', 'banana', 'cherry']
    another_list_instance = MyList(another_list)
    print(f"\nAnother original list: {another_list_instance.original_list}")
    print(f"Reversed another list: {another_list_instance.get_reversed_list()}")
    print(f"Sorted another list: {another_list_instance.get_sorted_list()}")
    print(f"Generated random numbers for another list: {another_list_instance.generate_random_numbers_list(min_val=10, max_val=20)}")

    # Example with invalid input (will raise an error)
    # try:
    #     invalid_list = MyList("not a list")
    # except TypeError as e:
    #     print(f"\nError creating MyList: {e}")
```

```python
# Exercise 3: Restaurant Menu Manager

class MenuManager:
    """
    Manages a restaurant menu, allowing for adding, updating, and deleting dishes.
    """
    def __init__(self):
        """
        Instantiates the 'menu' attribute as a list of dictionaries,
        representing the current dishes on the menu.
        Each dish dictionary has keys: 'name', 'price', 'spice_level', 'gluten_free'.
        Spice levels: A = not spicy, B = a little spicy, C = very spicy
        """
        self.menu = [
            {"name": "Soup", "price": 10, "spice_level": "B", "gluten_free": False},
            {"name": "Hamburger", "price": 15, "spice_level": "A", "gluten_free": True},
            {"name": "Salad", "price": 18, "spice_level": "A", "gluten_free": False},
            {"name": "French Fries", "price": 5, "spice_level": "C", "gluten_free": False},
            {"name": "Beef Bourguignon", "price": 25, "spice_level": "B", "gluten_free": True}
        ]
        print("MenuManager initialized with current dishes.")

    def add_item(self, name, price, spice_level, gluten_free):
        """
        Adds a new dish to the menu.

        Args:
            name (str): The name of the dish.
            price (int/float): The price of the dish.
            spice_level (str): The spice level (A, B, C).
            gluten_free (bool): True if gluten-free, False otherwise.
        """
        # Check if item already exists (optional, but good practice)
        for dish in self.menu:
            if dish["name"].lower() == name.lower():
                print(f"Dish '{name}' already exists in the menu. Consider updating it instead.")
                return

        new_dish = {
            "name": name,
            "price": price,
            "spice_level": spice_level.upper(), # Ensure consistent case for spice level
            "gluten_free": gluten_free
        }
        self.menu.append(new_dish)
        print(f"Dish '{name}' added to the menu.")

    def update_item(self, name, price, spice_level, gluten_free):
        """
        Updates an existing dish in the menu. Notifies if the dish is not found.

        Args:
            name (str): The name of the dish to update.
            price (int/float): The new price of the dish.
            spice_level (str): The new spice level (A, B, C).
            gluten_free (bool): The new gluten-free status.
        """
        found = False
        for dish in self.menu:
            if dish["name"].lower() == name.lower():
                dish["price"] = price
                dish["spice_level"] = spice_level.upper()
                dish["gluten_free"] = gluten_free
                found = True
                print(f"Dish '{name}' updated successfully.")
                break
        if not found:
            print(f"Dish '{name}' not found in the menu. Cannot update.")

    def remove_item(self, name):
        """
        Removes a dish from the menu. Notifies if the dish is not found.

        Args:
            name (str): The name of the dish to remove.
        """
        initial_menu_length = len(self.menu)
        # Use a list comprehension to create a new list excluding the item
        self.menu = [dish for dish in self.menu if dish["name"].lower() != name.lower()]

        if len(self.menu) < initial_menu_length:
            print(f"Dish '{name}' removed from the menu.")
            # print("Updated menu:", self.menu) # Uncomment to see the full updated list
        else:
            print(f"Dish '{name}' not found in the menu. Cannot remove.")

    def display_menu(self):
        """
        Prints the current menu in a readable format.
        """
        print("\n--- Current Menu ---")
        if not self.menu:
            print("The menu is currently empty.")
            return

        print(f"{'Name':<20} {'Price':<8} {'Spice':<8} {'Gluten-Free':<12}")
        print("-" * 50)
        for dish in self.menu:
            name = dish.get("name", "N/A")
            price = dish.get("price", "N/A")
            spice = dish.get("spice_level", "N/A")
            gluten = "Yes" if dish.get("gluten_free", False) else "No"
            print(f"{name:<20} {price:<8} {spice:<8} {gluten:<12}")
        print("-" * 50)


# --- Example Usage ---
if __name__ == "__main__":
    manager = MenuManager()

    # Display initial menu
    manager.display_menu()

    # Add new items
    print("\n--- Adding Items ---")
    manager.add_item("Pizza", 12, "A", False)
    manager.add_item("Tacos", 8, "C", True)
    manager.add_item("Soup", 10, "B", False) # Attempt to add existing item

    manager.display_menu()

    # Update an item
    print("\n--- Updating Items ---")
    manager.update_item("Hamburger", 16, "B", False)
    manager.update_item("Sushi", 20, "A", False) # Attempt to update non-existent item

    manager.display_menu()

    # Remove items
    print("\n--- Removing Items ---")
    manager.remove_item("Salad")
    manager.remove_item("Non-existent Dish") # Attempt to remove non-existent item

    manager.display_menu()
```
