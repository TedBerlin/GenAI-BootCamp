```python
# Exercise 1: Call History

class Phone:
    """
    Represents a phone with a number, call history, and message history.
    Allows making calls and sending/receiving messages.
    """
    def __init__(self, phone_number):
        """
        Initializes a Phone object.

        Args:
            phone_number (str): The phone number of this phone.
        """
        self.phone_number = phone_number
        self.call_history = []  # Stores strings like "X called Y"
        self.messages = []      # Stores dictionaries for messages

    def call(self, other_phone):
        """
        Simulates a call from this phone to another Phone object.
        Records the call in both phones' call history.

        Args:
            other_phone (Phone): The Phone object being called.
        """
        call_string = f"{self.phone_number} called {other_phone.phone_number}"
        self.call_history.append(call_string)
        other_phone.call_history.append(call_string) # Add to other phone's history too
        print(call_string)

    def show_call_history(self):
        """
        Prints the call history of this phone.
        """
        print(f"\n--- Call History for {self.phone_number} ---")
        if not self.call_history:
            print("No call history.")
        else:
            for call_record in self.call_history:
                print(call_record)
        print("-------------------------------------")

    def send_message(self, other_phone, content):
        """
        Sends a message from this phone to another Phone object.
        The message is saved in both sender's outgoing messages and receiver's incoming messages.

        Args:
            other_phone (Phone): The recipient Phone object.
            content (str): The content of the message.
        """
        message_record = {
            "to": other_phone.phone_number,
            "from": self.phone_number,
            "content": content
        }
        self.messages.append(message_record)
        other_phone._receive_message(message_record) # Add to recipient's messages
        print(f"Message sent from {self.phone_number} to {other_phone.phone_number}: '{content}'")

    def _receive_message(self, message_record):
        """
        Internal method to add a message to this phone's message history.
        Used by send_message from another Phone object.
        """
        self.messages.append(message_record)

    def show_outgoing_messages(self):
        """
        Prints messages sent from this phone.
        """
        print(f"\n--- Outgoing Messages for {self.phone_number} ---")
        outgoing_found = False
        for msg in self.messages:
            if msg["from"] == self.phone_number:
                print(f"To: {msg['to']}, Content: '{msg['content']}'")
                outgoing_found = True
        if not outgoing_found:
            print("No outgoing messages.")
        print("------------------------------------------")

    def show_incoming_messages(self):
        """
        Prints messages received by this phone.
        """
        print(f"\n--- Incoming Messages for {self.phone_number} ---")
        incoming_found = False
        for msg in self.messages:
            if msg["to"] == self.phone_number:
                print(f"From: {msg['from']}, Content: '{msg['content']}'")
                incoming_found = True
        if not incoming_found:
            print("No incoming messages.")
        print("------------------------------------------")

    def show_messages_from(self, other_phone_number):
        """
        Prints all messages (incoming and outgoing) associated with a specific other phone number.

        Args:
            other_phone_number (str): The phone number to filter messages by.
        """
        print(f"\n--- Messages with {other_phone_number} for {self.phone_number} ---")
        messages_found = False
        for msg in self.messages:
            if msg["from"] == other_phone_number:
                print(f"OUTGOING to {msg['to']}: '{msg['content']}'")
                messages_found = True
            elif msg["to"] == other_phone_number:
                print(f"INCOMING from {msg['from']}: '{msg['content']}'")
                messages_found = True
        if not messages_found:
            print(f"No messages with {other_phone_number}.")
        print("-------------------------------------------------")


# --- Test your code !!! ---
if __name__ == "__main__":
    # Instantiate Phone objects
    phone1 = Phone("0712345678")
    phone2 = Phone("0798765432")
    phone3 = Phone("0611223344")

    print("Phones created:")
    print(f"Phone 1: {phone1.phone_number}")
    print(f"Phone 2: {phone2.phone_number}")
    print(f"Phone 3: {phone3.phone_number}")

    # Test call method
    print("\n--- Testing Calls ---")
    phone1.call(phone2)
    phone2.call(phone3)
    phone3.call(phone1)
    phone1.call(phone3) # Another call from phone1 to phone3

    # Test show_call_history
    phone1.show_call_history()
    phone2.show_call_history()
    phone3.show_call_history()

    # Test send_message
    print("\n--- Testing Messages ---")
    phone1.send_message(phone2, "Hey Bob, how are you?")
    phone2.send_message(phone1, "I'm good, Alice! What's up?")
    phone1.send_message(phone3, "Hi Charlie, meeting at 3?")
    phone3.send_message(phone1, "Sounds good! See you then.")
    phone2.send_message(phone3, "Hello from Bob!")

    # Test show_outgoing_messages
    phone1.show_outgoing_messages()
    phone2.show_outgoing_messages()
    phone3.show_outgoing_messages()

    # Test show_incoming_messages
    phone1.show_incoming_messages()
    phone2.show_incoming_messages()
    phone3.show_incoming_messages()

    # Test show_messages_from
    phone1.show_messages_from("0798765432") # Messages between phone1 and phone2
    phone2.show_messages_from("0611223344") # Messages between phone2 and phone3
    phone3.show_messages_from("0712345678") # Messages between phone3 and phone1

    # Test with no history/messages
    phone4 = Phone("0700000000")
    phone4.show_call_history()
    phone4.show_outgoing_messages()
    phone4.show_incoming_messages()
    phone4.show_messages_from("0712345678")
```
