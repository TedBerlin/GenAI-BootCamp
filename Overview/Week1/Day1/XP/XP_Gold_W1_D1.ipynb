```python
# Exercise 1: What is the Season?

def get_season():
    """
    Asks the user for a month (1-12) and displays the corresponding season.
    """
    while True:
        try:
            month = int(input("Please enter a month as a number (1 to 12): "))
            if 1 <= month <= 12:
                break
            else:
                print("Invalid input. Please enter a number between 1 and 12.")
        except ValueError:
            print("Invalid input. Please enter a whole number.")

    season = ""
    if 3 <= month <= 5:
        season = "Spring"
    elif 6 <= month <= 8:
        season = "Summer"
    elif 9 <= month <= 11:
        season = "Autumn"
    elif month == 12 or 1 <= month <= 2:
        season = "Winter"
    else:
        # This case should ideally not be reached due to the input validation
        season = "Unknown Season"

    print(f"The month {month} falls in the season of: {season}")

# Call the function to run the program
if __name__ == "__main__":
    get_season()
```

```python
# Exercise 2: For Loop

# Part 1: Print all numbers from 1 to 20, inclusive.
print("--- Numbers from 1 to 20 ---")
for i in range(1, 21): # range(1, 21) generates numbers from 1 up to (but not including) 21
    print(i)

# Part 2: Print every number from 1 to 20 where the index is even.
# In Python, indices typically start from 0. If "index" refers to the number itself
# being even, it's straightforward. If it refers to its position in a 0-indexed sequence,
# we need to be careful. Assuming "index" refers to the number's value being even.
# If "index" means its position (0-indexed) in the loop, we would check `i % 2 == 0` for `range(20)`.
# Given the phrasing "where the index is even" for numbers from 1 to 20,
# it's more natural to interpret it as the number's value being even.

print("\n--- Even Numbers from 1 to 20 (where the number itself is even) ---")
for num in range(1, 21):
    if num % 2 == 0: # Check if the number is even
        print(num)

# Alternative interpretation for "index is even" if it refers to the loop's iteration index:
# This would print the 1st, 3rd, 5th, ... numbers (0-indexed: 0, 2, 4, ...)
# print("\n--- Numbers from 1 to 20 where the loop index is even (0-indexed) ---")
# for index, num in enumerate(range(1, 21)):
#     if index % 2 == 0:
#         print(num)
```

```python
# Exercise 3: While Loop

# Define your name (the name that will stop the loop)
# J'ai mis à jour le nom ici pour correspondre à "Thierry" comme vous l'avez indiqué.
MY_NAME = "Thierry"

print("--- Exercise 3: While Loop ---")
print(f"Please try to guess my name. The loop will stop when you enter '{MY_NAME}'.")

user_guess = "" # Initialize user_guess to an empty string

# Start the while loop
# The loop continues as long as user_guess is not equal to MY_NAME
while user_guess.strip().lower() != MY_NAME.lower(): # .strip() enlève les espaces, .lower() convertit en minuscules
    user_guess = input("Enter your guess for my name: ")
    # Appliquer les mêmes transformations à l'entrée de l'utilisateur pour une comparaison robuste
    if user_guess.strip().lower() != MY_NAME.lower():
        print("That's not it! Try again.")

print(f"Congratulations! You guessed my name: '{MY_NAME}'. The loop has stopped.")
```

```python
# Exercise 4: Check the Index

# Define the list of names
names = ['Samus', 'Cortana', 'V', 'Link', 'Mario', 'Cortana', 'Samus']

print("--- Exercise 4: Check the Index ---")
print(f"Here is our list of names: {names}")

# Ask the user for their name
user_name = input("Please enter a name to check its first occurrence in the list: ")

# Check if the user's name is in the list
if user_name in names:
    # If the name is in the list, print the index of its first occurrence
    # The .index() method returns the index of the first occurrence of the specified value.
    first_occurrence_index = names.index(user_name)
    print(f"The name '{user_name}' was found at the first occurrence index: {first_occurrence_index}")
else:
    # If the name is not in the list, inform the user
    print(f"The name '{user_name}' was not found in the list.")

```

```python
# Exercise 5: Greatest Number

def find_greatest_number():
    """
    Asks the user for three numbers and prints the greatest among them.
    """
    print("--- Exercise 5: Greatest Number ---")

    # Ask the user for the three numbers
    while True:
        try:
            num1 = float(input("Input the 1st number: "))
            break
        except ValueError:
            print("Invalid input. Please enter a valid number.")

    while True:
        try:
            num2 = float(input("Input the 2nd number: "))
            break
        except ValueError:
            print("Invalid input. Please enter a valid number.")

    while True:
        try:
            num3 = float(input("Input the 3rd number: "))
            break
        except ValueError:
            print("Invalid input. Please enter a valid number.")

    # Find the greatest number using built-in max() function
    greatest_number = max(num1, num2, num3)

    # Print the result
    print(f"The greatest number is: {greatest_number}")

# Call the function to run the program
if __name__ == "__main__":
    find_greatest_number()
```

```python
# Exercise 6: Random Number Game

import random

def play_guessing_game():
    """
    Allows the user to guess a random number between 1 and 9.
    Includes options to play again and tracks wins/losses.
    """
    print("--- Exercise 6: Random Number Game ---")
    print("Welcome to the Random Number Guessing Game!")

    total_games_won = 0
    total_games_lost = 0
    play_again = True

    while play_again:
        # Get a random number between 1 and 9 (inclusive)
        secret_number = random.randint(1, 9)
        guessed_correctly = False
        attempts = 0

        print("\nI'm thinking of a number between 1 and 9...")

        # Bonus 1: Loop for continuous guessing until correct or user quits current round
        while not guessed_correctly:
            attempts += 1
            user_guess_str = input(f"Attempt {attempts}: Enter your guess (1-9) or 'q' to quit this round: ").strip()

            if user_guess_str.lower() == 'q':
                print("Quitting this round.")
                break # Exit the inner guessing loop

            try:
                user_guess = int(user_guess_str)
                if not (1 <= user_guess <= 9):
                    print("Invalid input. Please enter a number between 1 and 9.")
                    continue # Ask again
            except ValueError:
                print("Invalid input. Please enter a whole number or 'q'.")
                continue # Ask again

            # Check the guess
            if user_guess == secret_number:
                print(f"Winner! You guessed the correct number ({secret_number}) in {attempts} attempts.")
                total_games_won += 1
                guessed_correctly = True
            else:
                print("Better luck next time. That's not the correct number.")
                if user_guess < secret_number:
                    print("Hint: Try a higher number.")
                else:
                    print("Hint: Try a lower number.")

        # If the user didn't guess correctly and didn't quit the round, it's a loss
        if not guessed_correctly and user_guess_str.lower() != 'q':
            print(f"You didn't guess the number. It was {secret_number}.")
            total_games_lost += 1

        # Ask if the user wants to play another game
        while True:
            play_choice = input("Do you want to play another game? (yes/no): ").strip().lower()
            if play_choice in ['yes', 'no']:
                break
            else:
                print("Invalid choice. Please enter 'yes' or 'no'.")

        if play_choice == 'no':
            play_again = False

    # Bonus 2: Display total games won and lost on exiting the loop
    print("\n--- Game Over ---")
    print(f"Total games played: {total_games_won + total_games_lost}")
    print(f"Games won: {total_games_won}")
    print(f"Games lost: {total_games_lost}")
    print("Thanks for playing!")

# Call the function to run the program
if __name__ == "__main__":
    play_guessing_game()
```
