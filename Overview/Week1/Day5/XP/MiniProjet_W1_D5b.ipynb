**Mini-Project - Tic Tac Toe**


```python
import os

# --- Global Variables (Game State) ---
# Represents the 3x3 Tic-Tac-Toe board.
# Using a list to represent the 9 squares. Index 0 is unused for 1-9 mapping.
board = [' ' for x in range(10)] # Index 0 is ignored, so board[1] to board[9] are used.
current_player = 'X' # Player 'X' starts the game.
game_on = True # Flag to control the game loop.

# ----------------------------------------------------------------------
## 1. Display the Board (`display_board()`)

#This function prints the current state of the Tic-Tac-Toe board to the console. It provides the visual representation for the players.

def display_board(board):
    """
    Clears the screen and displays the current Tic-Tac-Toe board.
    The board is a list of 10 strings, where index 0 is ignored.
    """
    os.system('cls' if os.name == 'nt' else 'clear') # Clears the console screen
    print('   |   |')
    print(' ' + board[7] + ' | ' + board[8] + ' | ' + board[9])
    print('   |   |')
    print('-----------')
    print('   |   |')
    print(' ' + board[4] + ' | ' + board[5] + ' | ' + board[6])
    print('   |   |')
    print('-----------')
    print('   |   |')
    print(' ' + board[1] + ' | ' + board[2] + ' | ' + board[3])
    print('   |   |')
    print("\n") # Add some space for readability
```

```python
# 2. Get Player Input (player_input(player))
#This function prompts the current player to choose a position on the board. It includes input validation to ensure the chosen position is valid (between 1 and 9) and empty.

def player_input(player):
    """
    Gets a valid position (1-9) from the current player.
    Ensures the input is a digit and the spot is available.
    """
    position = 0
    while position not in range(1, 10) or not is_space_free(board, position):
        try:
            position = int(input(f"Player {player}, choose your next position (1-9): "))
            if position not in range(1, 10):
                print("That is not a valid position. Please choose a number between 1 and 9.")
            elif not is_space_free(board, position):
                print("That spot is already taken! Please choose another one.")
        except ValueError:
            print("Invalid input. Please enter a number.")
    return position

def is_space_free(board, position):
    """Checks if a space on the board is empty."""
    return board[position] == ' '
```

```python
# 3. Place Mark on Board (place_mark(board, mark, position))
# This helper function updates the board with the player's mark at the chosen position.


def place_mark(board, mark, position):
    """
    Places the player's mark ('X' or 'O') on the board at the given position.
    """
    board[position] = mark
```

```python
# 4. Check Win Condition (check_win(board, mark))
# This function checks all possible winning combinations (rows, columns, and diagonals) to see if the current player has won.


def check_win(board, mark):
    """
    Checks if the current player (mark) has achieved a winning combination.
    Returns True if win, False otherwise.
    """
    # Check rows
    return ((board[1] == mark and board[2] == mark and board[3] == mark) or # Bottom row
            (board[4] == mark and board[5] == mark and board[6] == mark) or # Middle row
            (board[7] == mark and board[8] == mark and board[9] == mark) or # Top row
            # Check columns
            (board[1] == mark and board[4] == mark and board[7] == mark) or # Left column
            (board[2] == mark and board[5] == mark and board[8] == mark) or # Middle column
            (board[3] == mark and board[6] == mark and board[9] == mark) or # Right column
            # Check diagonals
            (board[1] == mark and board[5] == mark and board[9] == mark) or # Diagonal (left-bottom to right-top)
            (board[3] == mark and board[5] == mark and board[7] == mark))   # Diagonal (right-bottom to left-top)
```

```python
# 5. Check for Tie (check_tie(board))
# This function determines if the board is full, indicating a tie game if no winner has been found.

def check_tie(board):
    """
    Checks if all spaces on the board are taken, indicating a tie.
    Returns True if tie, False otherwise.
    """
    for i in range(1, 10):
        if is_space_free(board, i):
            return False # Found an empty space, so it's not a tie yet
    return True # All spaces are taken
```

```python
# 6. Switch Player (switch_player())
# This helper function toggles the current_player between 'X' and 'O' for the next turn.

def switch_player():
    """Switches the current player from 'X' to 'O' or 'O' to 'X'."""
    global current_player
    if current_player == 'X':
        current_player = 'O'
    else:
        current_player = 'X'
```

```python
# 7. Main Game Loop (play())
# This is the central function that orchestrates the entire game. It calls the other functions in sequence for each turn until the game ends (win or tie).

def play():
    """
    The main function to play a game of Tic-Tac-Toe.
    It manages the game flow, turns, and checks for win/tie conditions.
    """
    global game_on, board, current_player

    print("Welcome to Tic-Tac-Toe!")
    # Reset board for a new game
    board = [' ' for x in range(10)]
    current_player = 'X' # Always start with Player X
    game_on = True

    while game_on:
        # Display the board at the beginning of each turn
        display_board(board)

        # Get player's position
        position = player_input(current_player)

        # Place the mark on the board
        place_mark(board, current_player, position)

        # Check for a win
        if check_win(board, current_player):
            display_board(board) # Display final winning board
            print(f"Congratulations! Player {current_player} has won the game! 🎉")
            game_on = False # End the game
        # Check for a tie if no winner
        elif check_tie(board):
            display_board(board) # Display final tied board
            print("It's a tie! No one wins. 🤝")
            game_on = False # End the game
        # If no win and no tie, switch to the next player
        else:
            switch_player()

    # Ask if players want to play again
    play_again = input("Do you want to play again? (yes/no): ").lower()
    if play_again == 'yes':
        play() # Start a new game
    else:
        print("Thanks for playing! 👋")

# --- Start the game ---
if __name__ == "__main__":
    play()
```
