Étape 2 : Le Code de Traduction


```python
from googletrans import Translator

# La liste de mots français à traduire
french_words = ["Bonjour", "Au revoir", "Bienvenue", "A bientôt"]

# Initialiser le traducteur
translator = Translator()

# Créer un dictionnaire vide pour stocker les traductions
translated_dict = {}

# Parcourir chaque mot de la liste et le traduire
for word in french_words:
    # Traduire le mot du français ('fr') vers l'anglais ('en')
    # Le paramètre 'src' peut être omis pour la détection automatique,
    # mais il est bon de le spécifier pour plus de robustesse.
    # Le paramètre 'dest' est la langue de destination.
    translation = translator.translate(word, src='fr', dest='en')

    # Ajouter la paire mot original:traduction au dictionnaire
    translated_dict[word] = translation.text

# Afficher le résultat
print(translated_dict)
```

**Explication du Code**


* from googletrans import Translator: importe la classe Translator du module googletrans, qui est l'outil principal pour effectuer les traductions.

* french_words = [...]: liste de mots français fournie.

* translator = Translator(): instance de la classe Translator. C'est cet objet qui va gérer les requêtes de traduction.

* translated_dict = {}: dictionnaire vide est initialisé. C'est ici que nous stockerons les paires "mot français" : "traduction anglaise".

* for word in french_words:: itérations sur chaque word (mot français) dans la liste french_words.

* translation = translator.translate(word, src='fr', dest='en'):

> C'est la partie cruciale. La méthode translate() est appelée sur l'objet translator.

> word: texte à traduire.

> src='fr': Indique que la langue source est le français.

> dest='en': Indique que la langue de destination est l'anglais.

> La fonction translate() retourne un objet Translated qui contient diverses informations sur la traduction.

* translated_dict[word] = translation.text: accès à la traduction elle-même via l'attribut .text de l'objet translation et l'ajoutons à notre translated_dict avec le mot français original comme clé.

* print(translated_dict): le dictionnaire résultant est affiché.
