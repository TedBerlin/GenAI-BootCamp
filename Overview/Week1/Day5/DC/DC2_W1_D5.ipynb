Challenge 1 : Sorting

```python
# --- Programme de tri de mots ---

def sort_words_alphabetically():
    """
    Accepte une séquence de mots séparés par des virgules,
    les trie alphabétiquement et les affiche.
    """
    # 1. Demander à l'utilisateur de saisir la séquence de mots
    input_sequence = input("Veuillez entrer une séquence de mots séparés par des virgules : ")

    # 2. Séparer la chaîne en une liste de mots
    # La compréhension de liste est utilisée ici pour s'assurer qu'il n'y a pas d'espaces superflus autour des mots.
    words = [word.strip() for word in input_sequence.split(',')]

    # 3. Trier la liste de mots alphabétiquement
    # La fonction sorted() retourne une nouvelle liste triée,
    # sans modifier la liste originale 'words'.
    sorted_words = sorted(words)

    # 4. Rejoindre les mots triés en une chaîne séparée par des virgules
    output_sequence = ",".join(sorted_words)

    # 5. Afficher le résultat
    print(f"La séquence de mots triée est : {output_sequence}")

# Exécuter le programme
if __name__ == "__main__":
    sort_words_alphabetically()
```

Challenge 2 : Le Mot le Plus Long


```python
def longest_word(sentence):
    """
    Trouve le mot le plus long dans une phrase.
    Si plusieurs mots ont la même longueur maximale, retourne le premier.
    Les caractères de ponctuation sont comptés comme faisant partie du mot.

    Args:
        sentence (str): La phrase dans laquelle chercher le mot le plus long.

    Returns:
        str: Le mot le plus long trouvé.
    """
    # 1. Diviser la phrase en mots en utilisant les espaces comme délimiteurs.
    # Cela permet de conserver la ponctuation attachée aux mots.
    words = sentence.split()

    # 2. Initialiser le mot le plus long trouvé et sa longueur.
    # Nous commençons avec une chaîne vide, sa longueur sera 0.
    current_longest_word = ""
    max_length = 0

    # 3. Parcourir chaque mot dans la liste des mots
    for word in words:
        # 4. Calculer la longueur du word actuel
        word_length = len(word)

        # 5. Comparer la longueur du word actuel avec la longueur maximale trouvée jusqu'à présent.
        # Si le word actuel est plus long, ou s'il est le premier word (max_length est 0),
        # nous mettons à jour notre 'current_longest_word'.
        # L'ordre d'itération assure que le "premier" mot le plus long est conservé en cas d'égalité.
        if word_length > max_length:
            max_length = word_length
            current_longest_word = word

    # 6. Retourner le mot le plus long
    return current_longest_word

# --- Exemples d'utilisation et vérification ---
print("\"Margaret's toy is a pretty doll.\" ➞ \"{}\"".format(longest_word('Margaret\'s toy is a pretty doll.')))
print("\"A thing of beauty is a joy forever.\" ➞ \"{}\"".format(longest_word('A thing of beauty is a joy forever.')))
print("\"Forgetfulness is by all means powerless!\" ➞ \"{}\"".format(longest_word('Forgetfulness is by all means powerless!')))
print("\"Bonjour, monde!\" ➞ \"{}\"".format(longest_word('Bonjour, monde!')))
print("\"Un deux trois.\" ➞ \"{}\"".format(longest_word('Un deux trois.')))
```
