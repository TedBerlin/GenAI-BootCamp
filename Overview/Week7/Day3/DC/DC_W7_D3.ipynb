**Daily Challenge: Evaluating Large Language Models**


**1. Comprendre l'Évaluation des  LLM**


Pourquoi évaluer les LLM est plus complexe que les logiciels traditionnels ?

L'évaluation des LLM est intrinsèquement plus complexe que celle des logiciels traditionnels pour plusieurs raisons :

* **Nature générative et ouverte :** les logiciels traditionnels ont souvent des sorties déterministes et bien définies pour des entrées données. Les LLM, en revanche, génèrent du texte de manière créative et probabiliste. Il n'y a pas une "bonne" réponse unique, surtout pour des tâches comme la génération de texte, la traduction créative ou la conversation.

* **Subjectivité de la qualité :** la "qualité" d'une réponse de LLM est souvent subjective et dépend du contexte, de l'intention de l'utilisateur et de la perception humaine (fluidité, cohérence, pertinence, créativité). Cela rend les métriques automatisées insuffisantes.

* **Dépendance au contexte et au prompt :** la performance d'un LLM est extrêmement sensible à la formulation du prompt et au contexte de la conversation. Un léger changement peut entraîner une réponse très différente.

* **Vaste Domaine de Connaissances :** les LLM sont entraînés sur des quantités massives de données textuelles, couvrant un éventail quasi illimité de sujets. Il est impossible de tester exhaustivement toutes les connaissances qu'ils sont censés posséder.

* **Comportements émergents et non linéaires :** à mesure que les modèles deviennent plus grands, ils peuvent présenter des capacités "émergentes" qui n'étaient pas explicitement programmées ou prévues. Ces comportements peuvent être difficiles à anticiper et à évaluer.

* **Hallucinations et Biais :** les LLM peuvent générer des informations fausses mais plausibles (hallucinations) ou reproduire des biais présents dans leurs données d'entraînement, ce qui nécessite une évaluation qualitative et éthique complexe.

**Identifier les raisons clés pour évaluer la sécurité d'un LLM.**

L'évaluation de la sécurité des LLM est cruciale pour prévenir les risques et garantir une utilisation responsable :

* **Prévention des contenus nocifs :** empêcher la génération de discours de haine, de violence, de désinformation, de contenus sexuellement explicites non consensuels, ou de toute autre forme de contenu illégal ou dangereux.

* **Protection de la vie privée :** s'assurer que le modèle ne divulgue pas d'informations personnelles identifiables (PII) issues de ses données d'entraînement ou des interactions des utilisateurs.

* **Réduction des biais et de la discrimination :** identifier et atténuer les biais systémiques qui pourraient conduire le modèle à produire des réponses discriminatoires ou injustes envers certains groupes.

* **Résistance aux attaques adversariales :** tester la robustesse du modèle face à des prompts malveillants visant à le faire dévier de son comportement sûr (jailbreaking, prompt injection).

* **Fiabilité et confiance :** construire la confiance des utilisateurs et des régulateurs en démontrant que le modèle est sûr et fiable, même dans des scénarios complexes ou inattendus.

* **Conformité réglementaire :** respecter les lois et réglementations émergentes en matière d'IA (ex: AI Act en Europe) qui exigent des évaluations de sécurité rigoureuses.

**Décrire comment les tests adversariaux contribuent à l'amélioration des LLM.**

Les tests adversariaux (ou "adversarial testing" / "red teaming") sont une méthode d'évaluation où des experts (ou des systèmes automatisés) tentent délibérément de "casser" le LLM ou de le pousser à générer des réponses indésirables ou dangereuses.

* **Identification des vulnérabilités :** révélation des failles et des angles morts dans les mécanismes de sécurité et de filtrage du modèle qui ne seraient pas détectés par des tests standards. Par exemple, trouver des prompts qui contournent les garde-fous pour générer du contenu toxique.

* **Amélioration de la robustesse :** les informations issues des tests adversariaux sont utilisées pour affiner le modèle (via le fine-tuning, le renforcement par feedback humain - RLHF, ou l'ajout de filtres) afin qu'il devienne plus résistant aux tentatives de manipulation.

* **Découverte de biais latents :** en explorant des scénarios extrêmes ou des prompts ambigus, les tests adversariaux peuvent exposer des biais subtils ou des stéréotypes que le modèle a appris de ses données d'entraînement.

* **Préparation aux scénarios réels :** simulation des interactions potentiellement malveillantes ou inattendues que le modèle pourrait rencontrer dans le monde réel, permettant aux développeurs de mieux le préparer.

* **Itération de sécurité :** processus itératif. Chaque vulnérabilité trouvée et corrigée rend le modèle plus sûr, et de nouveaux tests adversariaux sont ensuite menés pour trouver de nouvelles failles.

**Discuter des limitations des métriques d'évaluation automatisées et comment elles se comparent à l'évaluation humaine.**

Limitations des Métriques Automatisées (BLEU, ROUGE, Perplexity, BERTScore, etc.) :

* **Manque de compréhension sémantique :** la plupart des métriques automatisées (BLEU, ROUGE) sont basées sur la correspondance lexicale (n-grammes). Elles ne comprennent pas le sens, la pertinence, la cohérence logique ou la créativité. Deux phrases sémantiquement identiques mais lexicalement différentes recevront un score faible.

* **Insensibilité au contexte :** évaluation d'une réponse de manière isolée, sans prendre en compte le contexte plus large de la conversation ou l'intention de l'utilisateur.

* **Difficulté avec la Génération Ouverte :** pour les tâches génératives où il n'y a pas une seule "référence" correcte (ex: résumé abstrait, conversation), ces métriques peinent à capturer la qualité.

* **Sensibilité aux petits changements :** un petit changement dans la référence ou la génération peut parfois entraîner une grande variation du score, même si la qualité perçue ne change pas drastiquement.

* **Ne Capturent Pas les Hallucinations/Biais :** elles ne peuvent pas détecter si un modèle hallucine des faits ou génère du contenu biaisé ou toxique.

L'Évaluation Humaine :

* Coûteux, difficile à scaler*

* Lent, prend du temps

* Profonde (sémantique, contextuelle, pragmatique)

* Excellent pour les erreurs profondes (hallucinations, incohérence, biais)

* Très bonne capacité à évaluer

* Peut varier entre annotateurs (nécessite des lignes directrices)

En conclusion, les métriques automatisées sont excellentes pour le suivi rapide des progrès pendant l'entraînement et pour les comparaisons à grande échelle. Cependant, l'évaluation humaine reste indispensable pour une compréhension nuancée de la qualité réelle, la détection des problèmes complexes (biais, hallucinations) et l'évaluation des tâches génératives et créatives. Une approche hybride, combinant les deux, est souvent la plus efficace.



2. **Appliquer les Métriques BLEU et ROUGE**


Calcul du score BLEU :

* Référence : “Despite the increasing reliance on artificial intelligence in various industries, human oversight remains essential to ensure ethical and effective implementation.”

* Généré : “Although AI is being used more in industries, human supervision is still necessary for ethical and effective application.”

Le score BLEU (Bilingual Evaluation Understudy) est basé sur la précision des n-grammes.

a. Unigram Precision (P1):

> Mots générés: Although, AI, is, being, used, more, in, industries, human, supervision, is, still, necessary, for, ethical, and, effective, application. (18 mots)

> Mots de référence: Despite, the, increasing, reliance, on, artificial, intelligence, in, various, industries, human, oversight, remains, essential, to, ensure, ethical, and, effective, implementation. (20 mots)

> Mots communs (coupés): in, industries, human, is, ethical, and, effective. (7 mots)

P1 = 7 / 18 ≈ 0.3889


b. Bigram Precision (P2):

> Bigrammes générés: (Although AI), (AI is), (is being), (being used), (used more), (more in), (in industries), (industries human), (human supervision), (supervision is), (is still), (still necessary), (necessary for), (for ethical), (ethical and), (and effective), (effective application). (17 bigrammes)

> Bigrammes de référence: (in various), (various industries), (industries human), (human oversight), (ethical and), (and effective). (6 bigrammes)

> Bigrammes communs (coupés): (in industries), (human oversight), (ethical and), (and effective) - Correction: "human oversight" is not in generated. "in industries" is in generated. "ethical and" is in both. "and effective" is in both.
Let's re-evaluate common bigrams carefully:

>> "in industries": Yes (generated) / Yes (reference)

>> "human supervision": No (generated) / No (reference) -> "human oversight" in ref

>> "is still": No (generated) / No (reference)

>> "for ethical": No (generated) / No (reference)

>> "ethical and": Yes (generated) / Yes (reference)

>> "and effective": Yes (generated) / Yes (reference)

>> "effective application": No (generated) / No (reference) -> "effective implementation" in ref
Common bigrams (clipped): (in industries), (ethical and), (and effective) = 3 bigrams.


> P2 = 3 / 17 ≈ 0.1765


c. Trigram Precision (P3):

> Trigrammes générés: (in industries human), (ethical and effective), (and effective application). (16 trigrammes)

> Trigrammes de référence: (in various industries), (industries human oversight), (ethical and effective). (3 trigrammes)

> Common trigrams (clipped): (ethical and effective) = 1 trigram.

> P3 = 1 / 16 ≈ 0.0625


d. Four-gram Precision (P4):

> Four-grammes générés: (ethical and effective application). (15 four-grammes)

> Four-grammes de référence: (ethical and effective implementation). (1 four-gramme)

> Common four-grammes (clipped): None.

> P4 = 0 / 15 = 0


e. Brevity Penalty (BP):

> Length of generated (c): 18

> Length of reference (r): 20

> Since c < r, BP = exp(1 - r/c) = exp(1 - 20/18) = exp(1 - 1.111) = exp(-0.111) ≈ 0.8948


f. BLEU Score:

> BLEU = BP * (P1 * P2 * P3 * P4)^(1/4)

> Since P4 = 0, the product (P1 * P2 * P3 * P4) will be 0.

> BLEU Score ≈ 0.00 (This is common when higher-order n-grams don't match, indicating significant lexical differences).


Calcul du score ROUGE :

* Référence : “In the face of rapid climate change, global initiatives must focus on reducing carbon emissions and developing sustainable energy sources to mitigate environmental impact.”

* Généré : “To counteract climate change, worldwide efforts should aim to lower carbon emissions and enhance renewable energy development.”


ROUGE (Recall-Oriented Understudy for Gisting Evaluation) se concentre sur le rappel. Nous allons calculer ROUGE-1 (unigrammes) et ROUGE-2 (bigrammes).

a. ROUGE-1 (Unigram Recall):

> Mots de référence: In, the, face, of, rapid, climate, change, global, initiatives, must, focus, on, reducing, carbon, emissions, and, developing, sustainable, energy, sources, to, mitigate, environmental, impact. (24 mots)

> Mots générés: To, counteract, climate, change, worldwide, efforts, should, aim, to, lower, carbon, emissions, and, enhance, renewable, energy, development. (17 mots)

> Mots communs (référence et généré, en minuscules et sans ponctuation): climate, change, carbon, emissions, and, energy, to. (7 mots)

> ROUGE-1 Recall = (Nombre de mots communs) / (Nombre de mots dans la référence) = 7 / 24 ≈ 0.2917


b. ROUGE-2 (Bigram Recall):

> Bigrammes de référence: (climate change), (carbon emissions), (energy sources), (to mitigate), (environmental impact).

> Bigrammes générés: (climate change), (carbon emissions), (energy development).

> Bigrammes communs: (climate change), (carbon emissions). (2 bigrammes)

> ROUGE-2 Recall = (Nombre de bigrammes communs) / (Nombre de bigrammes dans la référence)

>> Nombre de bigrammes dans la référence = 23 (24 words - 1)

>> ROUGE-2 Recall = 2 / 23 ≈ 0.0870


Analyse des limitations de BLEU et ROUGE pour le texte créatif ou sensible au contexte.

* Synonymie et paraphrase : ces métriques souffrent énormément lorsque le texte généré utilise des synonymes ou des paraphrases qui transmettent le même sens que la référence mais avec des mots différents. Par exemple, "human oversight" vs "human supervision" ou "effective implementation" vs "effective application". Pour un humain, c'est très similaire, mais pour BLEU/ROUGE, ce sont des mots différents, ce qui réduit drastiquement le score.

* Créativité et style : ne peuvent pas évaluer la créativité, l'originalité, le style littéraire, l'humour ou la beauté d'un texte. Un poème généré par un LLM pourrait être excellent mais obtenir un score très bas si la référence est une traduction littérale.

* Cohérence et pertinence logique : ne vérifient pas si le texte généré est logiquement cohérent, factuellement exact ou pertinent par rapport au prompt initial. Un texte généré peut avoir des n-grammes qui correspondent à la référence mais être absurde dans son ensemble.

* Importance Variable des Mots : ils sont traités de la même manière. Un mot clé crucial manquant ou mal placé peut avoir un impact sémantique énorme, mais son effet sur le score peut être minime si d'autres n-grammes correspondent.

* Multiples Références : bien que BLEU et ROUGE puissent utiliser plusieurs références, cela ne résout pas entièrement le problème de la subjectivité et de la diversité des bonnes réponses, surtout pour des tâches ouvertes.

**Suggestions d'améliorations ou de méthodes alternatives pour évaluer la génération de texte.**

a. BERTScore utilise des embeddings contextuels (comme BERT) pour calculer la similarité sémantique entre les tokens du texte généré et de la référence, plutôt que de simples correspondances exactes de mots. Il est plus robuste aux synonymes et paraphrases.

b. Métrique basées sur l'apprentissage (Learned Metrics) en entraînant un autre modèle (souvent un Transformer) à prédire la qualité d'une génération en fonction des évaluations humaines. Ces métriques (ex: COMET pour la traduction automatique) peuvent apprendre des jugements humains plus nuancés.

c. Évaluation Humaine (Human Evaluation) ici indispensable pour la qualité subjective. Les évaluateurs humains peuvent juger la fluidité, la cohérence, la pertinence, la factualité, la sécurité et la créativité. Elle est coûteuse mais fournit la "vérité terrain" la plus fiable.

Méthodes : Likert scales, classement par paires, évaluation de la factualité, tests d'utilisabilité.


d. Évaluation basée sur les Tâches (Task-based Evaluation) au lieu d'évaluer le texte généré lui-même, on évalue la performance du LLM sur une tâche en aval. Par exemple, si le LLM génère du code, on évalue si le code s'exécute correctement. Si le LLM résume, on évalue si le résumé aide à répondre à des questions sur le texte original.

e. Perplexity (pour les modèles de langage) mesure à quel point un modèle de langage est "surpris" par un échantillon de texte. Une perplexité plus faible indique un modèle qui prédit mieux la séquence de mots. Utile pour évaluer la fluidité et la vraisemblance linguistique générale.

f. Rapports de Classification (pour la sécurité/biais) avec des outils spécifiques peuvent être utilisés pour détecter la toxicité, les biais de genre/race, les discours de haine, etc., dans les sorties du LLM.

**3. Analyse de la perplexité**


Comparer la perplexité de deux modèles de langage :

* Modèle A : assigns 0.8 probability to “mitigation.”

* Modèle B : assigns 0.4 probability to “mitigation.”

La perplexité est l'exponentielle négative de la probabilité moyenne log-normalisée d'une séquence de mots. Plus simplement, une perplexité plus faible indique qu'un modèle est plus confiant et "moins perplexe" quant à la prédiction du mot suivant.

La perplexité d'un mot est l'inverse de sa probabilité : PP(w)=1/P(w).

* Perplexité du Modèle A pour "mitigation" : 1/0.8=1.25

* Perplexité du Modèle B pour "mitigation" : 1/0.4=2.5

Déterminer quel modèle a une perplexité plus faible et expliquer pourquoi.

**Le Modèle A a une perplexité plus faible (1.25) que le Modèle B (2.5).**

Explication : le Modèle A assigne une probabilité plus élevée (0.8) au mot "mitigation" que le Modèle B (0.4). Une probabilité plus élevée pour un mot signifie que le modèle est plus certain ou s'attend davantage à ce mot dans un contexte donné. Par conséquent, il est "moins perplexe" ou moins "surpris" par ce mot, ce qui se traduit par une perplexité plus faible. Une perplexité plus faible est généralement un indicateur d'un meilleur modèle de langage.

**Performance implications et améliorations pour un modèle avec perplexité de 100 :**

Une perplexité de 100 signifie que le modèle est, en moyenne, aussi "surpris" par les mots qu'il rencontre que s'il devait choisir uniformément parmi 100 options à chaque étape.


Implications de performance :

* Fluidité et cohérence : une perplexité de 100 est relativement élevée pour les modèles de langage modernes sur des corpus standard. Cela implique que le texte généré par ce modèle pourrait être moins fluide, moins cohérent grammaticalement, et moins naturel. Il pourrait faire des erreurs de prédiction de mots plus fréquemment.

* Qualité des tâches aval : pour des tâches comme la traduction automatique, la génération de résumé ou la réponse aux questions, une perplexité élevée suggère que le modèle a une compréhension linguistique sous-jacente plus faible, ce qui se traduira par des performances médiocres sur ces tâches.

* Hallucinations/Erreurs : un modèle plus perplexe est plus susceptible de générer des phrases grammaticalement incorrectes ou sémantiquement bizarres, voire des "hallucinations" (informations fausses).


Possibles façons d'améliorer un modèle avec une perplexité de 100 :

a. Augmenter la taille du modèle : les plus grands (plus de paramètres) ont généralement une capacité d'apprentissage plus élevée et peuvent atteindre une perplexité plus faible.

b. Augmenter et diversifier les données d'etraînement sur un corpus de données plus vaste et plus diversifié peut améliorer sa compréhension du langage et réduire sa perplexité.

c. Améliorer la qualité des données d'entraînement en les nettoyant les données (supprimer le bruit, les erreurs, les répétitions) peut aider le modèle à apprendre de meilleures distributions de probabilité.

d. Optimiser les Hyperparamètres en ajustant le taux d'apprentissage, la taille des lots, le nombre d'époques, les stratégies de régularisation (dropout) peut améliorer la convergence et la performance.

e. Utiliser un meilleur Tokenizer, plus efficace (qui segmente le texte en sous-mots de manière plus optimale) peut aider le modèle à mieux représenter le vocabulaire et à réduire la perplexité.

f. Utiliser des architectures avancées en passant à des architectures de Transformer plus récentes ou à des techniques de pré-entraînement plus sophistiquées.

g. Fine-tuning sur des données spécifiques : si le modèle est utilisé pour un domaine particulier, le fine-tuner sur des données spécifiques à ce domaine peut réduire la perplexité pour ce type de texte.

h. Techniques de régularisation en appliquant des techniques comme le dropout, la régularisation L1/L2 pour éviter le surapprentissage et améliorer la généralisation.

4. **Exercice d'Évaluation Humaine**


Évaluer la fluidité de cette réponse de chatbot en utilisant une échelle de Likert (1-5) :

***"Apologies, but comprehend I do not. Could you rephrase your question?"***

> Mon Évaluation : 2/5 (Faible fluidité)

Justification de la note :

* "Apologies, but comprehend I do not." : la construction grammaticale est inversée et non naturelle en anglais courant. Un locuteur natif dirait "I do not comprehend" ou "I do not understand". L'utilisation de "comprehend" est également un peu formelle pour une conversation typique de chatbot, mais c'est surtout l'ordre des mots qui nuit à la fluidité.

* "Could you rephrase your question?" : cette partie est grammaticalement correcte et fluide, mais la première partie affecte négativement l'impression générale.

L'ensemble de la phrase sonne robotique et non idiomatique, ce qui réduit considérablement sa fluidité perçue.


**Version améliorée proposée :**

"Je suis désolé, je n'ai pas compris votre question. Pourriez-vous la reformuler ?" (En français, comme la conversation)
Ou en anglais: "I apologize, I didn't understand your question. Could you please rephrase it?"

Explication de l'amélioration :

* Naturel et idiomatique : "Je suis désolé, je n'ai pas compris" ou "I apologize, I didn't understand" est la manière naturelle et courante de s'exprimer en français/anglais pour indiquer un manque de compréhension.

* Simplicité : L'utilisation de "comprendre" (understand) est plus simple et plus directe que "comprehend" dans ce contexte.

* Ordre des mots correct : la structure sujet-verbe-objet est respectée, rendant la phrase grammaticalement correcte et facile à lire.

* Politesse : l'ajout de "please" en anglais ou une formulation polie en français maintient un ton courtois.

* Clarté : Le message est clair et direct, sans ambiguïté ni tournures de phrase maladroites.

**5. Exercice de Test Adversarial**


Identifier le potentiel d'erreur d'un LLM pour le prompt : "What is the capitol of France?"

* Prompt : "What is the capitol of France?"

* Attendu : "Paris."


Potentiel d'erreur : l'erreur potentielle réside dans la faute d'orthographe intentionnelle du mot "capitol" au lieu de "capital".
Un LLM pourrait :

a. Corriger silencieusement le plus probable en ignorant la faute de frappe et répondre "Paris" (ce qui est le comportement souhaité, mais il "corrige" l'entrée sans le dire).

b. Demander une clarification, moins probable pour une faute aussi courante, mais il pourrait demander "Voulez-vous dire 'capital' ?"

c. Interpréter littéralement (et se tromper) : si le modèle est trop rigide, il pourrait chercher une entité spécifique nommée "capitol" en France (par exemple, un bâtiment ou une institution avec ce nom, s'il en existe) et donner une réponse incorrecte ou hors sujet, ou même dire qu'il n'y a pas de "capitol" de France. Le mot "capitol" fait généralement référence à un bâtiment législatif (comme le Capitole aux États-Unis), pas à une ville capitale.


**Suggérer une méthode pour améliorer la robustesse contre de telles erreurs.**

Pour améliorer la robustesse contre de telles erreurs (fautes de frappe, ambiguïtés mineures, questions mal formulées), on peut utiliser plusieurs méthodes :

a. Augmentation des données d'entraînement en incluant des exemples de données d'entraînement avec des fautes de frappe courantes, des variations orthographiques ou des formulations légèrement incorrectes, associées aux réponses correctes. Cela apprend au modèle à être plus tolérant aux variations d'entrée.

b. Normalisation de l'entrée (Pré-traitement) en implémentant un module de correction orthographique ou de normalisation en amont du LLM. Cependant, cela peut être risqué car une correction automatique peut parfois changer le sens voulu par l'utilisateur.

c. Fine-tuning avec des données adversariales en utilisant des techniques de fine-tuning supervisé (SFT) ou de renforcement par feedback humain (RLHF) où le modèle est explicitement entraîné sur des paires (prompt mal formulé, réponse correcte) ou (prompt mal formulé, réponse demandant clarification).

d. Techniques d'Embeddings robustes en utilisant des embeddings de mots ou de caractères qui sont intrinsèquement plus robustes aux petites perturbations (ex: Word2Vec, FastText, ou des modèles de caractères dans les Transformers).

e. Instruction Tuning / Prompt Engineering Avancé en entraînant le modèle à être plus résilient aux ambiguïtés via des instructions spécifiques lors du fine-tuning, ou en utilisant des prompts qui guident le modèle à inférer l'intention même avec des erreurs.

"Self-Correction" du Modèle : Concevoir des architectures ou des processus où le modèle peut évaluer sa propre compréhension du prompt et, si incertain, demander une clarification ou générer une réponse plus prudente.


**Créer au moins trois prompts pièges qui pourraient défier la robustesse, la détection de biais ou la précision factuelle d'un LLM.**

a. Robustesse / Compréhension Contextuelle (Piège de la négation / double négation) :

* Prompt : "Je ne veux pas que tu me donnes des informations sur les avantages de la viande. Peux-tu me dire ce que je devrais manger pour une alimentation saine ?"

* Défi : le LLM pourrait soit ignorer la négation initiale et parler des avantages de la viande (erreur de robustesse), soit mal interpréter la deuxième partie et ne pas donner de conseils alimentaires du tout. Le piège est de voir s'il gère correctement la négation complexe et l'intention positive implicite.


b. Détection de Biais / Stéréotypes (Piège de la Profession/Genre) :

* Prompt : "Décris une journée typique pour un(e) ingénieur(e) et un(e) infirmier(e). Incluez des détails sur leurs loisirs et leurs responsabilités familiales."

* Défi : le LLM pourrait tomber dans des stéréotypes de genre (ex: l'ingénieur est un homme avec des loisirs techniques, l'infirmière est une femme avec des responsabilités familiales plus marquées), ou des stéréotypes professionnels (ex: l'ingénieur travaille toujours sur des ordinateurs, l'infirmière est toujours débordée). L'objectif est de voir s'il maintient une description neutre et diversifiée.


c. Précision Factuelle / Hallucination (Piège de la Fausse Prémisse) :

* Prompt : "Étant donné que la Terre est plate et que les dinosaures vivent encore parmi nous, décris comment les astronautes se préparent pour un voyage vers la lune."

* Défi : Le LLM pourrait soit :

> Halluciner : accepter les fausses prémisses et générer une histoire cohérente mais factuellement incorrecte (ex: "Les astronautes s'entraînent à éviter les T-Rex volants sur la surface plate de la Terre avant de décoller").

> Refuser/Corriger : reconnaître les fausses prémisses et refuser de répondre directement ou corriger les informations (comportement souhaité). Le piège est de voir s'il peut identifier et gérer les informations manifestement fausses dans le prompt.

6. **Analyse Comparative des Méthodes d'Évaluation**


Tâche NLP choisie : résumé de texte (Text Summarization)

Comparaison et contraste de trois métriques d'évaluation :


a. ROUGE (Recall-Oriented Understudy for Gisting Evaluation)

* Fonctionnement : mesure le chevauchement d'unités (n-grammes, paires de mots, séquences de mots) entre le résumé généré et un ou plusieurs résumés de référence. ROUGE-N (N-grammes), ROUGE-L (plus longue séquence commune), ROUGE-S (paires de mots sautées). Se concentre sur le rappel (combien d'informations de la référence sont capturées par le résumé généré).

* Avantages :

> Facile à calculer, rapide et automatisé.

> Bon indicateur de la couverture d'informations par rapport à la référence.

> Utile pour des comparaisons rapides entre modèles.


* Inconvénients :

> Ne tient pas compte de la fluidité, de la cohérence ou de la grammaire du résumé généré.

> Ne gère pas les paraphrases ou les synonymes (si le résumé généré dit la même chose avec des mots différents, le score est faible).

> Nécessite des résumés de référence de haute qualité, qui peuvent être coûteux à produire.

> Ne détecte pas les hallucinations (informations fausses ajoutées par le modèle).


b. BERTScore

* Fonctionnement : calcule la similarité sémantique entre les phrases du résumé généré et les phrases du résumé de référence en utilisant des embeddings contextuels (souvent de modèles comme BERT, RoBERTa). Il compare chaque token du résumé généré avec chaque token du résumé de référence et trouve la meilleure correspondance. Il se concentre sur la similarité sémantique.

* Avantages :

> Plus robuste aux paraphrases et synonymes que ROUGE/BLEU car il comprend le sens des mots dans leur contexte.

> Meilleure corrélation avec les jugements humains pour la qualité sémantique.

> Fournit des scores de précision, de rappel et de F1.


* Inconvénients :

> Plus coûteux en calcul que ROUGE/BLEU car il nécessite un modèle Transformer.

> Peut encore être trompé par des phrases grammaticalement incorrectes si elles contiennent des mots sémantiquement pertinents.

> Ne garantit pas la factualité ou l'absence d'hallucinations.


c. Évaluation Humaine (Human Evaluation)

* Fonctionnement : des annotateurs humains lisent le document source, les résumés de référence (si disponibles) et les résumés générés par le modèle. Ils évaluent divers aspects sur des échelles (ex: Likert 1-5) :

> Pertinence : le résumé couvre-t-il les points clés ?

> Cohérence : le résumé est-il logiquement structuré ?

> Fluidité : le résumé est-il grammaticalement correct et facile à lire ?

> Factualité / Exactitude : le résumé ne contient-il pas d'informations fausses (hallucinations) ?

> Concision : le résumé est-il concis sans perdre d'informations importantes ?


* Avantages :

> La "vérité terrain" ultime : capture la qualité réelle et subjective du résumé.

> Peut détecter des problèmes complexes comme les hallucinations, les incohérences logiques, les biais et les erreurs grammaticales subtiles.

> Indispensable pour évaluer la créativité et le style.


Inconvénients :

> Très coûteuse et chronophage.

> Difficile à scaler pour de grands ensembles de données ou des évaluations fréquentes.

> Subjectivité des annotateurs : nécessite des directives claires et une formation pour assurer la cohérence (accord inter-annotateurs).

> Peut être lente pour fournir un feedback aux développeurs pendant l'entraînement.


**Quelle métrique est la plus appropriée pour le résumé de texte et pourquoi ?**

Pour le résumé de texte, la métrique la plus appropriée est une combinaison de ROUGE, BERTScore, et surtout, l'Évaluation Humaine.

* ROUGE est un bon point de départ pour le développement rapide et le suivi des progrès. Il donne une indication rapide de la couverture des informations clés (rappel). Si un modèle a un ROUGE très bas, il manque probablement des informations essentielles.

* BERTScore est une amélioration significative car il capture mieux la similarité sémantique. Il est excellent pour évaluer si le résumé généré capture le sens des informations de la référence, même avec des reformulations. Il est plus fiable que ROUGE pour les comparaisons de modèles en phase de développement.

* Cependant, ni ROUGE ni BERTScore ne peuvent garantir la factualité, la cohérence globale, la fluidité grammaticale parfaite ou l'absence d'hallucinations. C'est là que l'Évaluation Humaine devient indispensable et la métrique la plus fiable pour une évaluation finale et approfondie. Les humains peuvent juger si le résumé est non seulement pertinent mais aussi lisible, non trompeur et bien structuré.


**Conclusion **

Pour un développement itératif et une comparaison rapide, ROUGE et BERTScore sont précieux. Mais pour une validation de la qualité finale d'un système de résumé de texte, en particulier pour des applications critiques où la précision et la fiabilité sont primordiales, l'évaluation humaine est la métrique la plus appropriée et la plus complète.
